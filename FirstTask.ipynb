```python
import sympy as sp
```

```python
# переменные и функции для примеров
x = sp.symbols("x")
y = sp.Function("y")
```

```python
equation = sp.Eq(sp.diff(y(x), x, 2) + y(x), 0)
```

```python
solution = sp.dsolve(
    equation,
    ics={
        y(0): 1,
        sp.diff(y(x), x).subs(x, 0): 1
    }
)

solution
```
