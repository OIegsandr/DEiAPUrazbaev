```python
import sympy as sp
```

```python
# переменные и функции для примеров
x = sp.symbols("x")
y = sp.Function("y")
```

```python
equation = sp.Eq(((1 - x**2)) * (sp.diff(y(x), x, 1)), 2 * x * y(x))
```

```python
solution = sp.dsolve(
    equation,
    ics={
        y(0): 1,
        sp.diff(y(x), x).subs(x, 0): 1
    }
)

solution
```
