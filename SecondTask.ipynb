```python
import sympy as sp
```

```python
# переменные и функции для примеров
x = sp.symbols("x")
y = sp.Function("y")
z = sp.symbols("z")
```

```python
equationB = sp.Eq(((1 - x**2)) * (sp.diff(y(x), x, 1)), 2 * x * y(x))
```

```python
solutionB = sp.dsolve(
    equationB,
    ics={
        y(0): 1,
        sp.diff(y(x), x).subs(x, 0): 1
    }
)

solutionB
```
